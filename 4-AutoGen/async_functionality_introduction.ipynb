```python
## Synchronous or Normal way of Coding

import time 

def brew_coffee():
    print("Brewing Coffee")
    time.sleep(3)
    print("Coffee Ready")

def toast_bagel():
    print("Toasting bagel")
    time.sleep(2) 
    print("Bagel Ready")


def main():
    start = time.time()

    brew_coffee()
    toast_bagel()

    end = time.time()

    print(f"Time:{end-start:.2f} minutes")

main()
```

```python
## Asyncronous Version
import time 
import asyncio

async def brew_coffee():
    print("Brewing Coffee")
    await asyncio.sleep(3) 
    print("Coffee Ready")

async def toast_bagel():
    print("Toasting bagel")
    await asyncio.sleep(2)
    print("Bagel Ready")



async def main():
    start = time.time()
    coffee = brew_coffee()
    bagel = toast_bagel()

    await asyncio.gather(coffee, bagel)

    end = time.time()

    print(f"Time:{end-start:.2f} minutes")


await main()
```
